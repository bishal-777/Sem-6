## Scalar, Vector, Matrix and Tensors

#### Scalar- 0D tensors

```python
torch.tensor(5)
```

#### Vector- 1D tensors

```python
torch.tensor([1, 2, 3, 4, 5])
```

#### Matrix- 2D tensors

```python
torch.tensor([[3, 2, 3],
             [2, 5, 0],
             [4, 8, 9]])
```

#### Tensors- ND tensors(order >=3)

```python
torch.tensor([[[3, 2, 3],
             [2, 5, 0],
             [4, 8, 9]]])
```

## Pytorch tensor and Numpy arrays 

```python
import numpy as np

array = np.array([1, 2, 3, 4, 5])
array
 
```

```python
tensor = torch.tensor([4, 1, 5, 4, 5])
tensor
```

#### Converting numpy array to pytorch tensor

```python
array_to_tensor = torch.from_numpy(array)
array_to_tensor
```

#### Converting pytorch tensor to numpy array

```python
tensor_to_array = tensor.numpy()
tensor_to_array
```
