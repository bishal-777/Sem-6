````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
import torch
import torch.nn as nn
````

## Task-1

````python
# Define model
class LinearRegressionNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(
            in_features=1,
            out_features=1,
            bias=True
        )

    def forward(self, x):
        return self.linear(x)
````

Using the model definition above, determine whether the model is capable of learning from each dataset provided in Lab 4:

linear-regression-data1.csv

assignment-data.csv

assignment-data2.csv

Note: Only determine whether the model can be trained on the datasets and whether its architecture

is appropriate. A detailed performance evaluation is not required

````python
df1 = pd.read_csv("../datasets/assignment-data.csv")
df2 = pd.read_csv("../datasets/assignment-data2.csv")
df3 = pd.read_csv("../datasets/linear-regression-data1.csv")
````

````python
X1 = torch.tensor(df1['x'].values, dtype=torch.float32)
y1 = torch.tensor(df1['y'].values, dtype=torch.float32)

X2 = torch.tensor(df2['x'].values, dtype=torch.float32)
y2 = torch.tensor(df2['y'].values, dtype=torch.float32)

X3 = torch.tensor(df3['x'].values, dtype=torch.float32)
y3 = torch.tensor(df3['y'].values, dtype=torch.float32)



print(f"Data 1(assignment-data.csv): {X1.shape, y1.shape}")
print(f"Data 2(assignment-dat2.csv): {X2.shape, y2.shape}")
print(f"Data 3(linear-regression-data1.csv): {X3.shape, y3.shape}")
````

````python
X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size=0.2, random_state=42)
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.2, random_state=42)    
X3_train, X3_test, y3_train, y3_test = train_test_split(X3, y3, test_size=0.2, random_state=42)
````

````python
# Scatter plot
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.scatter(df1.iloc[:, 0], df1.iloc[:, 1])
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Dataset-1 Visualization")

plt.subplot(1, 3, 2)
plt.scatter(df2.iloc[:, 0], df2.iloc[:, 1])
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Dataset-2 Visualization")

plt.subplot(1, 3, 3)
plt.scatter(df3.iloc[:, 0], df3.iloc[:, 1])
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Dataset-3 Visualization")

plt.tight_layout()
plt.show()
````

````python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.scatter(X1_train, y1_train, label='Training Data', alpha=0.7)
plt.scatter(X1_test, y1_test, label='Testing Data', alpha=0.7)
plt.title('Dataset-1: Training vs. Testing Split')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.grid(True)

plt.subplot(1, 3, 2)
plt.scatter(X2_train, y2_train, label='Training Data', alpha=0.7)
plt.scatter(X2_test, y2_test, label='Testing Data', alpha=0.7)
plt.title('Dataset-2: Training vs. Testing Split')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.grid(True)

plt.subplot(1, 3, 3)
plt.scatter(X3_train, y3_train, label='Training Data', alpha=0.7)
plt.scatter(X3_test, y3_test, label='Testing Data', alpha=0.7)
plt.title('Dataset-3: Training vs. Testing Split')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
````

### Among the three datasets, the model is capable of learning from the 3rd dataset(linear-regression-data1.csv) because third dataset follows a straight line patterns and other datasets follow quadratic patterns.
